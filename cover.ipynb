```python
from manim import *
```

```python
scale = 6
step = 1

config.background_color = WHITE
config.frame_width = scale * 2
config.pixel_width = 1920
config.pixel_height = 1920
```

```python
def createNumberPlane(self):
    self.camera.background_color = WHITE
       
    return NumberPlane(
        x_range = (-scale * 2, scale * 2, step),
        y_range = (-scale * 2, scale * 2, step),
        axis_config ={  "color": "#808080", "stroke_width": scale * 2 },
        background_line_style = { "stroke_color": WHITE, "stroke_width": 0 }
    )
```

```python
%%manim -v WARNING Folium

# from ./algebra
class Folium(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)

        folium = ImplicitFunction(
            lambda x, y: x ** 3 + y ** 3 - 10 * x * y,
            color = BLACK,
            stroke_width = 18,
            x_range = [-scale - 5, scale + 5],
            y_range = [-scale - 5, scale + 5],
        )

        self.add(numberplane, folium)
```

```python
%%manim -v WARNING VertStretch

class VertStretch(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)

        cubed = FunctionGraph(
            lambda t: (t - 3) * t * (t + 3) * (1/9),
            color = BLACK,
            stroke_width = 18
        )

        strech = FunctionGraph(
            lambda t: 2 * ((t - 3) * t * (t + 3) * (1/9)),
            x_range=[-6, 6],
            color = BLACK,
            stroke_width=18
        )
        dashed = DashedVMobject(strech, num_dashes=60, equal_lengths =True)

        self.add(numberplane, cubed, dashed)
```

```python
%%manim -v WARNING Monomial

class Monomial(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)

        func = FunctionGraph(lambda x: .5 * x**2, color=BLACK, stroke_width=18)

        self.add(numberplane, func)
```

```python
%%manim -v WARNING Logistic 

class Logistic(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)

        func = FunctionGraph(lambda x: 4/(1 + 10**(-.5 * x)), color=BLACK, stroke_width=18)
        
        self.add(numberplane, func)
```

```python
%%manim -v WARNING Polynomial

class Polynomial(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)

        func = FunctionGraph(lambda t: (t - 3) * t * (t + 3) * (1/6), color=BLACK, stroke_width=18)
    
        self.add(numberplane, func)
```

```python
%%manim -v WARNING Rational

class Rational(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)

        func = FunctionGraph(
            lambda x: .5/x,
            color = BLACK,
            stroke_width=18,
            discontinuities=[0],  # discontinuous points
            dt=0.01
        )

        self.add(numberplane, func)
```

```python
%%manim -v WARNING SineLemniscate

def func(theta):
    r = np.sqrt(np.sin(2 * theta))
    return np.array([r * np.cos(theta) * 6, r * np.sin(theta) * 6, 0])


# from ./polar
class SineLemniscate(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)

        curl1 = ParametricFunction(
            func, 
            t_range = np.array([0+0.000001, (PI/2) - 0.0000001]), 
            fill_opacity=0, 
            stroke_width=18
        ).set_color(BLACK)

        curl2 = ParametricFunction(
            func, 
            t_range = np.array([PI+0.0000001, (3 * PI/2) - 0.0000001]), 
            fill_opacity=0, 
            stroke_width=18
        ).set_color(BLACK)

        self.add(numberplane, curl1, curl2)
```

```python
%%manim -v WARNING Parametric

def func(theta):
    return np.array([np.sin(4 * theta) * 5, np.cos(6 * theta) * 5 , 0]) 


class Parametric(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)

        curl = ParametricFunction(func, t_range = np.array([0, 2 * TAU]), fill_opacity=0, stroke_width=18).set_color(BLACK)

        self.add(numberplane, curl)
```

```python
%%manim -v WARNING Step

class Step(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)
        
        steps = VGroup()
        steps.add(Line(start=numberplane.c2p(-6, -3), end=numberplane.c2p(0, -3), color=BLACK, stroke_width=18))
        steps.add(Line( start=numberplane.c2p(0, -3), end=numberplane.c2p(8, 5), color=BLACK, stroke_width=18))
        steps.add(Dot(numberplane.c2p(0, -3), color=BLACK, radius=0.24))
    
        self.add(numberplane, steps)
```

```python
%%manim -v WARNING Step

class Step(Scene):
    def construct(self):
        numberplane = createNumberPlane(self)
        
        steps = VGroup()
        filled_dots = VGroup()
        hollow_dots = VGroup()

        steps.add(Line(start=numberplane.c2p(-3, -3), end=numberplane.c2p(0, -3), color=BLACK, stroke_width=18))
        filled_dots.add(Dot(numberplane.c2p(-3, -3), color=BLACK, radius=0.24))
        hollow_dots.add(Dot(numberplane.c2p(0, -3), color=BLACK, radius=0.24))
        hollow_dots.add(Dot(numberplane.c2p(0, -3), color=WHITE, radius=0.12))
        
        steps.add(Line(start=numberplane.c2p(0, 3), end=numberplane.c2p(3, 3), color=BLACK, stroke_width=18))
        
        filled_dots.add(Dot(numberplane.c2p(0, 3), color=BLACK, radius=0.24))
        hollow_dots.add(Dot(numberplane.c2p(3, 3), color=BLACK, radius=0.24))
        hollow_dots.add(Dot(numberplane.c2p(3, 3), color=WHITE, radius=0.12))
    
        self.add(numberplane, steps, filled_dots, hollow_dots)
```

```python
%%manim -v WARNING GridTransformation

class GridTransformation(Scene):
    def construct(self):
        self.camera.background_color = WHITE
        numberplane =  NumberPlane(
            x_range = (-scale * 2, scale * 2, step),
            y_range = (-scale * 2, scale * 2, step),
            axis_config ={ "color": BLACK, "stroke_width": 0 },
            background_line_style = { "stroke_color": WHITE, "stroke_width": 0 }
        )
        
        axes = VGroup()
        axes.add(DashedVMobject(Line(
            start=numberplane.c2p(-6, 0),
            end=numberplane.c2p(6, 0),
            color=BLACK,
            stroke_width=12,
            stroke_opacity=.45
        )))
        
        axes.add(DashedVMobject(Line(
            start=numberplane.c2p(0, -6),
            end=numberplane.c2p(0, 6),
            color=BLACK,
            stroke_width=12,
            stroke_opacity=.45
        )))
        
        funcs = VGroup()
        
        funcs.add(FunctionGraph(lambda x: x/3, color=BLACK, stroke_width=9))
        funcs.add(FunctionGraph(lambda x: x/3 + (11/3), color=BLACK, stroke_width=9))
        funcs.add(FunctionGraph(lambda x: x/3 - (11/3), color=BLACK, stroke_width=9))
        
        funcs.add(FunctionGraph(lambda x: 4 * x, color=BLACK, stroke_width=9))
        funcs.add(FunctionGraph(lambda x: 4 * x - 11, color=BLACK, stroke_width=9))
        funcs.add(FunctionGraph(lambda x: 4 * x + 11, color=BLACK, stroke_width=9))

        

        poly = Polygon([0,0,0], [3, 1, 0], [4, 5, 0], [1, 4, 0], color=BLACK, stroke_width=18, fill_color=GREY_A, fill_opacity=1)

        arrows = VGroup()
        arrows.add(Arrow(start=[0, 0, 0],end=[3 * 1.07, 1 * 1.07, 0], color=BLACK, stroke_width=18))
        arrows.add(Arrow(start=[0, 0, 0],end=[1 * 1.07, 4 * 1.07, 0], color=BLACK, stroke_width=18))

        self.add(axes, funcs, poly, arrows)
```
