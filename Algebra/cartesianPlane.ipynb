```python
from manim import *
```

```python
scale = 6
step = 3

config.background_color = WHITE
config.frame_width = scale * 2
config.pixel_width = 1920
config.pixel_height = 1920
```

```python
def createNumberPlane(self):
    self.camera.background_color = WHITE

    return NumberPlane(
        x_range=(-scale * 2, scale * 2, step),
        y_range=(-scale * 2, scale * 2, step),
        axis_config={ "color": "#808080", "stroke_width": scale * 1.5},
        background_line_style={ "stroke_color": "#808080", "stroke_width": scale * 1, "stroke_opacity": .5 },
    )
```

```python
%%manim -v WARNING CartesianShape

class CartesianShape(Scene):
    def construct(self):
        numberplane =createNumberPlane(self)

        square = Polygon((1,1,0),(1,5,0),(5,5,0),(5,1,0), color=BLACK, stroke_width=18)
        
        point1 = Text("(1, 1)", color=BLACK, font_size=45).move_to([1, 1, 0] + DOWN * .5)
        point2 = Text("(3, 1)", color=BLACK, font_size=45).move_to([5, 1, 0] + DOWN * .5)
        point3 = Text("(1, 3)", color=BLACK, font_size=45).move_to([1, 5, 0] + UP * .5)
        point4 = Text("(3, 3)", color=BLACK, font_size=45).move_to([5, 5, 0] + UP * .5)

        self.add(numberplane, square, point1, point2, point3, point4)
```

```python
%%manim -v WARNING CartesianGraph

class CartesianGraph(Scene):
    def construct(self):
        numberplane =createNumberPlane(self)

        func = FunctionGraph(lambda x: 2 * x, color=BLACK, stroke_width=18)

        self.add(numberplane, func)
```

```python
%%manim -v WARNING CartesianQuadrants

class CartesianQuadrants(Scene):
    def construct(self):
        numberplane =createNumberPlane(self)

        
        point1 = Text("I", color=BLACK, font_size=150).move_to([3, 3, 0])
        point2 = Text("II", color=BLACK, font_size=150).move_to([-3, 3, 0])
        point3 = Text("III", color=BLACK, font_size=150).move_to([-3, -3, 0])
        point4 = Text("IV", color=BLACK, font_size=150).move_to([3, -3, 0])

        self.add(numberplane, point1, point2, point3, point4)
```
