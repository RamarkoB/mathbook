```python
from manim import *
import numpy as np
```

```python
scale = 3
step = 1

config.background_color = WHITE
config.frame_width = scale * 2
config.pixel_width = 1920
config.pixel_height = 1920
```

```python
%%manim -v WARNING VertStretch

config.frame_width = scale * 2

class VertStretch(Scene):
    def construct(self):
        self.camera.background_color = WHITE
        numberplane = NumberPlane(
            x_range = (-scale, scale, step),
            y_range = (-scale, scale, step),
            axis_config = {
                "color": BLACK, 
                "stroke_width": scale
            }, 
            background_line_style = {
                "stroke_color": WHITE,
                "stroke_width": 2,
                "stroke_opacity": 0.6
            })

        cubed = FunctionGraph(
            lambda t: (t - 1.5) * t * (t + 1.5),
            color = BLACK,
            stroke_width = 10
        )

        strech = FunctionGraph(
            lambda t: 2 * (t - 1.5) * t * (t + 1.5),
            x_range=[-3, 3],
            color = BLACK,
            stroke_width = 5,
        )
        dashed = DashedVMobject(strech, num_dashes=150)

        self.add(numberplane, cubed, dashed)
```

```python
%%manim -v WARNING VertShift

config.frame_width = scale * 2

class VertShift(Scene):
    def construct(self):
        self.camera.background_color = WHITE
        numberplane = NumberPlane(
            x_range = (-scale, scale, step),
            y_range = (-scale, scale, step),
            axis_config = {
                "color": BLACK, 
                "stroke_width": scale
            }, 
            background_line_style = {
                "stroke_color": WHITE,
                "stroke_width": 2,
                "stroke_opacity": 0.6
            })

        cubed = FunctionGraph(
            lambda t: (t - 1.5) * t * (t + 1.5),
            color = BLACK,
            stroke_width = 10
        )

        strech = FunctionGraph(
            lambda t: 1 + (t - 1.5) * t * (t + 1.5),
            x_range=[-3, 3],
            color = BLACK,
            stroke_width = 5,
        )
        dashed = DashedVMobject(strech, num_dashes=75)

        self.add(numberplane, cubed, dashed)
```

```python
%%manim -v WARNING HoriStretch

config.frame_width = scale * 2

class HoriStretch(Scene):
    def construct(self):
        self.camera.background_color = WHITE
        numberplane = NumberPlane(
            x_range = (-scale, scale, step),
            y_range = (-scale, scale, step),
            axis_config = {
                "color": BLACK, 
                "stroke_width": scale
            }, 
            background_line_style = {
                "stroke_color": WHITE,
                "stroke_width": 2,
                "stroke_opacity": 0.6
            })

        cubed = FunctionGraph(
            lambda t: (t - 1.5) * t * (t + 1.5),
            color = BLACK,
            stroke_width = 10
        )

        strech = FunctionGraph(
            lambda t: ((.5 * t) - 1.5) * (.5 * t) * ((.5 * t) + 1.5),
            x_range=[-3, 3],
            color = BLACK,
            stroke_width = 5,
        )
        dashed = DashedVMobject(strech, num_dashes=15)

        self.add(numberplane, cubed, dashed)
```

```python
%%manim -v WARNING HoriShift

config.frame_width = scale * 2

class HoriShift(Scene):
    def construct(self):
        self.camera.background_color = WHITE
        numberplane = NumberPlane(
            x_range = (-scale, scale, step),
            y_range = (-scale, scale, step),
            axis_config = {
                "color": BLACK, 
                "stroke_width": scale
            }, 
            background_line_style = {
                "stroke_color": WHITE,
                "stroke_width": 2,
                "stroke_opacity": 0.6
            })

        cubed = FunctionGraph(
            lambda t: (t - 1.5) * t * (t + 1.5),
            color = BLACK,
            stroke_width = 10
        )

        strech = FunctionGraph(
            lambda t: ((t - .5) - 1.5) * (t - .5) * ((t - .5) + 1.5),
            x_range=[-3, 3],
            color = BLACK,
            stroke_width = 5,
        )
        dashed = DashedVMobject(strech, num_dashes=100)

        self.add(numberplane, cubed, dashed)
```
